```python
# prompt: 针对身高(cm)及体重(kg)两个栏位，绘制散布图，x轴名称为height，y轴名称为weight

import pandas as pd
import matplotlib.pyplot as plt

# Install openpyxl if you haven't already
# !pip install openpyxl  # Uncomment if you haven't installed it

try:
    df = pd.read_excel("test.xlsx", engine='openpyxl')
    #print(df)

    # Create the scatter plot
    plt.figure(figsize=(8, 6))  # Adjust figure size if needed
    plt.scatter(df['身高(cm)'], df['体重(kg)'])
    plt.xlabel('height')
    plt.ylabel('weight')
    plt.title('Height vs. Weight Scatter Plot')
    plt.grid(True)  # Add a grid for better readability
    plt.show()

except FileNotFoundError:
    print("Error: test.xlsx not found.")
except KeyError as e:
    print(f"Error: Column '{e}' not found in the DataFrame. Please check your column names.")
except Exception as e:
    print(f"An error occurred: {e}")
```

```python
# prompt: 针对身高(cm)及体重(kg)两个栏位资料，以简单线性回归找出此函数方程式，谢谢

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

try:
    df = pd.read_excel("test.xlsx", engine='openpyxl')

    # Prepare the data
    X = df[['身高(cm)']]  # Features (height)
    y = df['体重(kg)']    # Target (weight)

    # Create and train the linear regression model
    model = LinearRegression()
    model.fit(X, y)

    # Get the coefficients (slope and intercept)
    slope = model.coef_[0]
    intercept = model.intercept_

    # Print the linear equation
    print(f"Linear Regression Equation: Weight = {slope:.2f} * Height + {intercept:.2f}")

    # Make predictions for plotting the regression line
    y_pred = model.predict(X)

    # Plot the data and the regression line
    plt.figure(figsize=(8, 6))
    plt.scatter(X, y, label='Data Points')
    plt.plot(X, y_pred, color='red', label='Regression Line')
    plt.xlabel('Height (cm)')
    plt.ylabel('Weight (kg)')
    plt.title('Height vs. Weight with Linear Regression')
    plt.legend()
    plt.grid(True)
    plt.show()

except FileNotFoundError:
    print("Error: test.xlsx not found.")
except KeyError as e:
    print(f"Error: Column '{e}' not found in the DataFrame. Please check your column names.")
except Exception as e:
    print(f"An error occurred: {e}")
```
