```python
# 安裝必要套件（如果尚未安裝）
# !pip install requests beautifulsoup4 pandas

# 匯入需要用到的套件
import requests
from bs4 import BeautifulSoup
import pandas as pd

# 目標網址
url = "https://woman.udn.com/woman/story/123164/8748702?list_ch2_index"

# 發送 GET 請求取得網頁內容
response = requests.get(url)

# 如果網頁回應正常（狀態碼為 200）
if response.status_code == 200:
    # 使用 BeautifulSoup 解析 HTML 原始碼
    soup = BeautifulSoup(response.text, "html.parser")

    # 取得所有 <h4> 標籤，回傳為列表
    h4_tags = soup.find_all("h4")

    # 建立空列表，準備儲存標籤與文字內容
    data = []

    # 逐一處理每個 h4 標籤
    for tag in h4_tags:
        tag_html = str(tag)              # 將整個 h4 標籤轉為字串（包含 <h4> 標籤本身）
        tag_text = tag.get_text(strip=True)  # 取得 h4 內的文字，並去除前後空白
        data.append({"Tag": tag_html, "Text": tag_text})  # 存入資料列表中

    # 將資料轉為 DataFrame 格式
    df = pd.DataFrame(data)

    # 顯示結果
    print("共找到", len(df), "個 <h4> 標籤：")
    display(df)  # Colab 中可視化顯示 DataFrame（在 Jupyter 類環境下使用）
else:
    # 若回應失敗則輸出錯誤訊息
    print(f"無法連線到網頁，HTTP 狀態碼：{response.status_code}")

```

```python
# prompt: 接下來我要取得全部a標籤且 class="site-product__link list-ellipsis"，并且以dataframe的方式輸出所有標籤内的文字及標籤内的·超鏈接，並保留第21個到32個，再將其顯示在熒幕和匯出成excel（僅呈現保留的標籤（# 保留第 21 個到第 32 個（索引從 20 到 31））  df_subset = df.iloc[20:32]），並重新在表格中重新計數

# 匯入需要用到的套件
import requests
from bs4 import BeautifulSoup
import pandas as pd

# 目標網址
url = "https://woman.udn.com/woman/story/123164/8748702?list_ch2_index"

# 發送 GET 請求取得網頁內容
response = requests.get(url)

# 如果網頁回應正常（狀態碼為 200）
if response.status_code == 200:
  # 使用 BeautifulSoup 解析 HTML 原始碼
  soup = BeautifulSoup(response.text, "html.parser")

  # 尋找所有符合條件的 a 標籤
  a_tags = soup.find_all("a", class_="site-product__link list-ellipsis")

  # 建立空列表，準備儲存文字內容和超鏈接
  data = []

  # 逐一處理每個 a 標籤
  for tag in a_tags:
    text = tag.get_text(strip=True)  # 取得標籤內的文字，並去除前後空白
    href = tag.get('href')  # 取得超鏈接（href 屬性）
    data.append({"Text": text, "Href": href})  # 存入資料列表中

  # 將資料轉為 DataFrame 格式
  df = pd.DataFrame(data)

  # 保留第 21 個到第 32 個（索引從 20 到 31）
  df_subset = df.iloc[20:32].copy()

  # 重新計算索引
  df_subset = df_subset.reset_index(drop=True)

  # 顯示結果
  print("保留的 a 標籤 for 生活娛樂 (第 21 到 32 個)：")
  display(df_subset)  # Colab 中可視化顯示 DataFrame

  # 匯出成 Excel 檔案
  df_subset.to_excel("a_tags_subset.xlsx", index=False)
  print("\n資料已匯出至 a_tags_subset.xlsx")

else:
  # 若回應失敗則輸出錯誤訊息
  print(f"無法連線到網頁，HTTP 狀態碼：{response.status_code}")
```
