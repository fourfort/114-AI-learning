```python
# 安裝 openai 套件，確保使用最新版本
!pip install --upgrade openai

# 載入所需的函式庫
import os
from openai import OpenAI

# 設定 OpenAI API 金鑰
API_KEY = "sk-AlmOZWbkYM88uQwliZC8UNxbN4mN4HqbMU5DXDU7Hf8TZA76"  # 替換你的 API Key
CUSTOM_HOST = "https://api.chatanywhere.org"  # 替换为您需要使用的 Host

# 实例化 OpenAI 客户端并設置自定義 Host
client = OpenAI(
    api_key=API_KEY,
    base_url=CUSTOM_HOST
)

def chat_with_gpt():
    print("輸入你的問題，輸入 'exit' 來結束對話。")

    while True:
        user_input = input("你: ")
        if user_input.lower() == "exit":
            print("結束對話。")
            break

        try:
            # 呼叫 OpenAI Chat API 生成回應
            response = client.chat.completions.create(
                model="gpt-4",  # 使用的模型
                messages=[
                    {"role": "system", "content": "你是個友善且有幫助的助手。"},
                    {"role": "user", "content": user_input}
                ]
            )
            # 取得模型回應
            reply = response.choices[0].message.content
            print(f"ChatGPT: {reply}")

        except Exception as e:
            print(f"發生錯誤: {e}")

# 執行對話
chat_with_gpt()
```
