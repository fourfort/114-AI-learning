```python
# prompt: 請提供Apriori演算法簡單的執行範例，不使用function，請舉10筆資料作爲範例進行操作 (運用pandas、mlxtend)
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori

# 範例資料 (10筆交易)
dataset = [
    ['牛奶', '啤酒', '尿布'],
    ['牛奶', '啤酒', '咖啡'],
    ['牛奶', '尿布', '啤酒', '雞蛋'],
    ['啤酒', '雞蛋'],
    ['牛奶', '尿布', '雞蛋'],
    ['尿布', '啤酒', '咖啡'],
    ['牛奶', '啤酒'],
    ['牛奶', '雞蛋'],
    ['啤酒', '尿布'],
    ['牛奶', '啤酒', '尿布', '咖啡']
]

dataset[1]

```

```python
# 使用 TransactionEncoder 將資料轉換成適合 Apriori 演算法的格式
te = TransactionEncoder()
te_ary = te.fit(dataset).transform(dataset)
df = pd.DataFrame(te_ary, columns=te.columns_)

```

```python
# 執行 Apriori 演算法，設定最小支持度 (min_support)
frequent_itemsets = apriori(df, min_support=0.5, use_colnames=True)
# 顯示結果(頻繁項集)
frequent_itemsets
```

```python
# prompt: 顯示所有頻繁集中，confidence >= 0.5之關聯規則

from mlxtend.frequent_patterns import association_rules

# 計算關聯規則，設定最小信賴度 (min_threshold)
rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.5)

# 顯示結果(關聯規則)
rules
```

```python
# prompt: 使用 DataFrame rules: 匯出成excel

# Export the DataFrame to an Excel file
rules.to_excel('rules.xlsx', index=False) # Replace 'rules.xlsx' with your desired filename
```
