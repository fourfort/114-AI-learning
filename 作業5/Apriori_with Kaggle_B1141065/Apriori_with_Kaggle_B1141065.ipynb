```python
# prompt: 請讀取Bakery.csv,並顯示前五筆資料

import pandas as pd

# 假設 Bakery.csv 存在於 Colab 的工作區中
df = pd.read_csv('Bakery.csv')

# 顯示前五筆資料
df.head()
```

```python
# prompt: 使用 DataFrame df: 請顯示敘述統計

# Display descriptive statistics for numerical columns
df.describe()
```

```python
# prompt: 請刪除同一個TransactionNo中，重複的items的筆數,刪除後存到新的變數df_g,再保留原本的欄位名稱

df_g = df.groupby(['TransactionNo', 'Items'])['Items'].count().reset_index(name='count')
df_g = df_g.drop_duplicates(subset=['TransactionNo', 'Items'], keep='first')
df_g = df_g[['TransactionNo','Items' , 'count' ]]
df_g
```

```python
# prompt: 請在一個TransactionNo中，刪除重複的其中一筆資料，刪除後存到新的變數df_g,保留原本的欄位名稱，產生此動作的程式碼即可

df_g = df.drop_duplicates(subset=['TransactionNo', 'Items'], keep='first')
df_g
```

```python
# prompt: 使用 DataFrame df_g: 轉成 Items group by TransactionNo

# Group the dataframe by 'TransactionNo' and aggregate the 'Items' column.
items_by_transaction = df_g.groupby('TransactionNo')['Items'].apply(list)

# Display the result (optional).
items_by_transaction
```

```python
# prompt: 請將items_by_transaction的資料轉成one-hot編碼格式 （運用 Transaction Encoder）

import pandas as pd
from mlxtend.preprocessing import TransactionEncoder

te = TransactionEncoder()
te_ary = te.fit(items_by_transaction).transform(items_by_transaction)
df_onehot = pd.DataFrame(te_ary, columns=te.columns_)
df_onehot
```

```python
# prompt: 進行Apriori產生頻繁集，mini support 設定為 0.05

from mlxtend.frequent_patterns import apriori

# 設定最小支持度
min_support = 0.03

# 使用 apriori 函數找出頻繁項目集
frequent_itemsets = apriori(df_onehot, min_support=min_support, use_colnames=True)

# 顯示頻繁項目集
frequent_itemsets
```

```python
# prompt: 產生關聯規則，信賴度設定爲0.1

from mlxtend.frequent_patterns import association_rules

# 設定最小信賴度
min_threshold = 0.01

# 使用 association_rules 函數找出關聯規則
rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=min_threshold)

# 顯示關聯規則
rules
```

```python
# prompt: 使用 DataFrame rules: 請依照lift由大到小來進行顯示

# Sort the dataframe by the 'lift' column in descending order.
rules_sorted = rules.sort_values('lift', ascending=False)

# Display the sorted dataframe.
rules_sorted
```

```python
# prompt: 使用 DataFrame rules_sorted: 請匯出成excel

# Export the dataframe to an Excel file
rules_sorted.to_excel('rules_sorted.xlsx', index=False)
```
